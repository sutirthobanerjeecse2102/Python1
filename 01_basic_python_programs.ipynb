## 1. Sum and average of the first 55 natural numbers

```python
total = 0

for number in range(1, 56):
    total += number

average = total / 55

print("Sum:", total)
print("Average:", average)
```

## 2. Factorial of a number (enter 7 for the assigned example)

```python
number = int(input("Enter a non-negative integer (7 for the assigned example): "))

if number < 0:
    print("Factorial is not defined for negative integers.")
else:
    factorial = 1

    for value in range(1, number + 1):
        factorial *= value

    print(f"{number}! =", factorial)
```

## 3. Reverse a string

```python
text = input("Enter a string: ")
reversed_text = text[::-1]

print("Reversed string:", reversed_text)
```

## 4. Check whether a year is a leap year

```python
year = int(input("Enter a year: "))

if year % 400 == 0:
    print(year, "is a leap year.")
elif year % 100 == 0:
    print(year, "is not a leap year.")
elif year % 4 == 0:
    print(year, "is a leap year.")
else:
    print(year, "is not a leap year.")
```
