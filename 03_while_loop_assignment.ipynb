## 1. Find the sum of the digits of an integer

```python
number = int(input("Enter an integer: "))
remaining = abs(number)
digit_sum = 0

while remaining > 0:
    digit = remaining % 10
    digit_sum += digit
    remaining //= 10

print("Sum of digits:", digit_sum)
```

## 2. Reverse an integer using a while loop

```python
number = int(input("Enter an integer: "))
remaining = abs(number)
reversed_number = 0

while remaining > 0:
    digit = remaining % 10
    reversed_number = reversed_number * 10 + digit
    remaining //= 10

if number < 0:
    reversed_number = -reversed_number

print("Reversed number:", reversed_number)
```

## 3. Keep accepting numbers until 0 and print their sum

```python
total = 0

while True:
    number = int(input("Enter a number (0 to stop): "))

    if number == 0:
        break

    total += number

print("Sum of entered numbers:", total)
```
