## 1. Tuple: remove duplicates while preserving original order

```python
original = (1, 2, 2, 3, 1, 4, 3)
unique_values = []

for value in original:
    if value not in unique_values:
        unique_values.append(value)

result = tuple(unique_values)
print("Original tuple:", original)
print("Tuple without duplicates:", result)
```

## 2. Tuple: sort students by marks in descending order

```python
students = (("Riya", 85), ("Aman", 92), ("Soham", 78), ("Priya", 89))

sorted_students = sorted(students, key=lambda student: student[1], reverse=True)

print("Students in descending order of marks:")
for name, marks in sorted_students:
    print(name, "-", marks)
```

## 3. Tuple: input five integers and calculate sum and average

```python
numbers = []

for index in range(5):
    number = int(input(f"Enter integer {index + 1}: "))
    numbers.append(number)

values = tuple(numbers)
total = sum(values)
average = total / len(values)

print("Tuple:", values)
print("Sum:", total)
print("Average:", average)
```

## 4. Tuple: count occurrences of 10

```python
values = (10, 20, 10, 30, 10, 40, 20)
count = 0

for value in values:
    if value == 10:
        count += 1

print("Tuple:", values)
print("Number of times 10 occurs:", count)
```

## 5. Tuple: find maximum and minimum without max() or min()

```python
values = (14, 3, 27, 8, 19, 2)

largest = values[0]
smallest = values[0]

for value in values[1:]:
    if value > largest:
        largest = value
    if value < smallest:
        smallest = value

print("Tuple:", values)
print("Largest element:", largest)
print("Smallest element:", smallest)
```

## 6. Set: display union, intersection and difference

```python
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

print("Union:", set_a | set_b)
print("Intersection:", set_a & set_b)
print("Difference (A - B):", set_a - set_b)
```

## 7. Set: find common distinct elements in two user-entered lists

```python
list_a = [int(value) for value in input("Enter the first list of integers, separated by spaces: ").split()]
list_b = [int(value) for value in input("Enter the second list of integers, separated by spaces: ").split()]

common_elements = set(list_a) & set(list_b)

print("Common distinct elements:", common_elements)
```

## 8. Set: check subset and superset relationships

```python
first_set = {1, 2, 3}
second_set = {1, 2, 3, 4, 5}

print("First set:", first_set)
print("Second set:", second_set)
print("Is the first set a subset of the second?", first_set.issubset(second_set))
print("Is the second set a superset of the first?", second_set.issuperset(first_set))
```

## 9. Set: find students enrolled in exactly one course

```python
python_students = {"Aman", "Riya", "Soham", "Priya"}
java_students = {"Riya", "Priya", "Rahul", "Ankit"}

exactly_one_course = python_students.symmetric_difference(java_students)

print("Students enrolled in exactly one course:", exactly_one_course)
```

## 10. Dictionary: store and display marks of five students

```python
student_marks = {}

for index in range(5):
    name = input(f"Enter the name of student {index + 1}: ")
    marks = float(input(f"Enter marks for {name}: "))
    student_marks[name] = marks

print("\nStudent marks:")
for name, marks in student_marks.items():
    print(f"{name}: {marks}")
```

## 11. Dictionary: count the frequency of each word in a sentence

```python
sentence = input("Enter a sentence: ").lower()
word_frequency = {}

for word in sentence.split():
    word = word.strip(".,!?;:\"'()[]{}")
    if word:
        word_frequency[word] = word_frequency.get(word, 0) + 1

print("Word frequencies:")
print(word_frequency)
```

## 12. Dictionary: merge two dictionaries and add common values

```python
dictionary_one = {"Pen": 10, "Pencil": 5, "Book": 3}
dictionary_two = {"Pen": 4, "Pencil": 7, "Eraser": 6}

merged_dictionary = dictionary_one.copy()

for item, quantity in dictionary_two.items():
    if item in merged_dictionary:
        merged_dictionary[item] += quantity
    else:
        merged_dictionary[item] = quantity

print("Merged dictionary:", merged_dictionary)
```
