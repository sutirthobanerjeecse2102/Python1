## 1. Arithmetic operations using separate functions

```python
def add(a, b):
    return a + b


def subtract(a, b):
    return a - b


def multiply(a, b):
    return a * b


def divide(a, b):
    if b == 0:
        return "Division by zero is not allowed."
    return a / b


first = float(input("Enter the first number: "))
second = float(input("Enter the second number: "))

print("Addition:", add(first, second))
print("Subtraction:", subtract(first, second))
print("Multiplication:", multiply(first, second))
print("Division:", divide(first, second))
```

## 2. Temperature conversion functions

```python
def celsius_to_fahrenheit(celsius):
    return (celsius * 9 / 5) + 32


def fahrenheit_to_celsius(fahrenheit):
    return (fahrenheit - 32) * 5 / 9


def celsius_to_kelvin(celsius):
    return celsius + 273.15


temperature = float(input("Enter the temperature: "))
print("Celsius to Fahrenheit:", celsius_to_fahrenheit(temperature))
print("Fahrenheit to Celsius:", fahrenheit_to_celsius(temperature))
print("Celsius to Kelvin:", celsius_to_kelvin(temperature))
```

## 3. Create and import a factorial module

```python
# Run this cell to create the module file named module_factorial.py.
module_code = """def factorial(number):
    if number < 0:
        raise ValueError('Factorial is not defined for negative integers.')

    result = 1
    for value in range(1, number + 1):
        result *= value
    return result
"""

with open("module_factorial.py", "w", encoding="utf-8") as module_file:
    module_file.write(module_code)

print("module_factorial.py has been created.")
```

## 5. Check whether a number is prime using a function

```python
def is_prime(number):
    if number < 2:
        return False

    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            return False

    return True


number = int(input("Enter an integer: "))

if is_prime(number):
    print(number, "is a prime number.")
else:
    print(number, "is not a prime number.")
```

## 6. Reverse a string using string_reverse()

```python
def string_reverse(text):
    return text[::-1]


user_text = input("Enter a string: ")
print("Reversed string:", string_reverse(user_text))
```
